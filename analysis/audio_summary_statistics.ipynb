```python
import os
import pandas as pd
from pydub.utils import mediainfo
import matplotlib.pyplot as plt
```

```python
def get_audio_lengths_recursively(folder_path):
    audio_lengths = []
    for root, dirs, files in os.walk(folder_path):
        for file in files:
            if file.endswith('.m4a'):  # Only process WAV files
                file_path = os.path.join(root, file)
                try:
                    info = mediainfo(file_path)
                    duration = float(info['duration'])  # Duration in seconds
                    relative_path = os.path.relpath(file_path, folder_path)
                    audio_lengths.append((relative_path, duration))
                except Exception as e:
                    print(f"Could not process file {file}: {e}")
    return audio_lengths

# Folder containing subfolders with audio files
folder_path = 'path_to_your_audio_folder'

# Get lengths of audio files
audio_files = get_audio_lengths_recursively(folder_path)

# Print results
for file, length in audio_files:
    print(f"{file}: {length:.2f} seconds")
```

```python
# Folder containing subfolders with audio files
folder_path = r"C:\Users\Yang Qing\OneDrive\Desktop\Evaluating LLM\Recording\Raw Transcription"

# Get lengths of audio files
audio_files = get_audio_lengths_recursively(folder_path)

```

```python
# Save results in a DataFrame
df = pd.DataFrame(audio_files, columns=["File", "Length (seconds)"])

# Convert length to minutes
df["Length (minutes)"] = df["Length (seconds)"] / 60

# Calculate the average length
average_length = df["Length (minutes)"].mean()
print(f"Average Length: {average_length:.2f} minutes")

# Plot histogram
plt.figure(figsize=(10, 6))
plt.hist(df["Length (minutes)"], bins=20, edgecolor='black', alpha=0.7)
plt.title("Distribution of Audio File Lengths (in minutes)")
plt.xlabel("Length (minutes)")
plt.ylabel("Frequency")
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```
